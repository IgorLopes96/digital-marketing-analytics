```python
import pandas as pd
df = pd.read_csv('/kaggle/input/datasets/rabieelkharoua/predict-conversion-in-digital-marketing-dataset/digital_marketing_campaign_dataset.csv')
df.head()
```

```python
df.describe()
```

## Customer Engagement Variables

The Following Variables were selected because they represent customer engagement and purchasing behavior:

- EmailOpens
- EmailClicks
- WebsiteVisits
- PagesPerVisits
- TimeOnSite
- LoyaltyPoints
- PreviousPurchases

These variables will be used to identify customer segments.


```python
engagement_columns = [
    'EmailOpens',
    'EmailClicks',
    'WebsiteVisits',
    'PagesPerVisit',
    'TimeOnSite',
    'LoyaltyPoints',
    'PreviousPurchases'
    
]

df[engagement_columns].describe()
```

```python
df['EngagementScore'] = (
    df['EmailOpens'] +
    df['EmailClicks'] +
    df ['WebsiteVisits'] +
    df['PagesPerVisit'] +
    df['TimeOnSite']
)

df['EngagementScore'].describe()
```

```python
df['EngagementSegment'] = pd.qcut (
    df['EngagementScore'],
    q=3,
    labels=[
    'Low Engagement',
    'Medium Engagement',
    'High Engagement'
    ]
)
df['EngagementSegment'].value_counts()
```

```python
df.groupby('EngagementSegment')['Conversion'].mean()
```

```python
df.groupby('EngagementSegment')[[
    'EmailOpens',
    'EmailClicks',
    'WebsiteVisits',
    'PagesPerVisit',
    'TimeOnSite'
]].mean()
```

### Low engagement Segment

- Customers: 2,667
- Conversion Rate: 80.8%
- Lowest email interaction
- Lowest website activity

Business risk:

These customers are less engaged with marketing content and have the lowest conversion performance.

Recommended Action:

Launch re-engagement campaigns to increase interaction and conversion potential.

```python
low_engagement = df[
df['EngagementSegment'] == 'Low Engagement'
]
low_engagement.head()
```

```python
low_engagement.to_csv(
    'low_engagement_customers.csv',
    index=False
)
```

# Key Findings

- Customers were segmented into Low, Medium, and High Engagement Groups.
- High Engagement customers achieved the highest conversion rate (92.2%).
- Medium engagement customers achieved a conversion rate of 89.9%
- Low Engagement customers achieved the lowest conversion rate (80.8%)
- Higher engagement levels were associated with higher conversion rates.


# Recommendations

- Prioritize conversion campaigns for High Engagement customers.
- Develop personalized campaigns for Medium Engagement customers.
- Create re-engagement initiatives for low-engagement customers.
- Focus on increasing email interaction and website engagement. 

# Conclusion

The analysis identified a clear relationship between customer engagement and conversion performance.

Cusotmers with higher engagement levels consistently achieved higher engagement levels consistently achieved higher conversion rates. These findings support the use of engagement-based segmentation to improve marketing effectiveness and customer retention.
